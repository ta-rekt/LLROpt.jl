### Pong

```python
import pygame
import random

# Initialize pygame
pygame.init()

# Set up the game window
WIDTH, HEIGHT = 800, 600
window = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption('Pong')

# Set up colors
WHITE = (255, 255, 255)
BLACK = (0, 0, 0)

# Set up the game clock
clock = pygame.time.Clock()

# Paddle properties
PADDLE_WIDTH = 15
PADDLE_HEIGHT = 100
PADDLE_SPEED = 10

# Ball properties
BALL_SIZE = 15
BALL_SPEED_X = 5
BALL_SPEED_Y = 5

# Font for score display
font = pygame.font.SysFont('arial', 30)

# Function to draw paddles, ball, and scores
def draw_window(left_paddle, right_paddle, ball, left_score, right_score):
    window.fill(BLACK)
    # Draw paddles
    pygame.draw.rect(window, WHITE, pygame.Rect(left_paddle[0], left_paddle[1], PADDLE_WIDTH, PADDLE_HEIGHT))
    pygame.draw.rect(window, WHITE, pygame.Rect(right_paddle[0], right_paddle[1], PADDLE_WIDTH, PADDLE_HEIGHT))
    # Draw ball
    pygame.draw.ellipse(window, WHITE, pygame.Rect(ball[0], ball[1], BALL_SIZE, BALL_SIZE))
    # Draw mid line
    pygame.draw.aaline(window, WHITE, (WIDTH // 2, 0), (WIDTH // 2, HEIGHT))
    # Draw scores
    left_score_text = font.render(str(left_score), True, WHITE)
    right_score_text = font.render(str(right_score), True, WHITE)
    window.blit(left_score_text, (WIDTH // 4 - left_score_text.get_width() // 2, 20))
    window.blit(right_score_text, (WIDTH * 3 // 4 - right_score_text.get_width() // 2, 20))
    # Update display
    pygame.display.update()

# Main game function
def main():
    left_paddle = [50, HEIGHT // 2 - PADDLE_HEIGHT // 2]
    right_paddle = [WIDTH - 50 - PADDLE_WIDTH, HEIGHT // 2 - PADDLE_HEIGHT // 2]
    ball = [WIDTH // 2 - BALL_SIZE // 2, HEIGHT // 2 - BALL_SIZE // 2]
    ball_velocity = [BALL_SPEED_X * random.choice((1, -1)), BALL_SPEED_Y * random.choice((1, -1))]

    left_score = 0
    right_score = 0

    running = True
    while running:
        clock.tick(60)  # Set the frame rate to 60 FPS
        
        # Event handling
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False

        # Paddle movement (up and down)
        keys = pygame.key.get_pressed()
        if keys[pygame.K_w] and left_paddle[1] > 0:
            left_paddle[1] -= PADDLE_SPEED
        if keys[pygame.K_s] and left_paddle[1] < HEIGHT - PADDLE_HEIGHT:
            left_paddle[1] += PADDLE_SPEED
        if keys[pygame.K_UP] and right_paddle[1] > 0:
            right_paddle[1] -= PADDLE_SPEED
        if keys[pygame.K_DOWN] and right_paddle[1] < HEIGHT - PADDLE_HEIGHT:
            right_paddle[1] += PADDLE_SPEED

        # Ball movement
        ball[0] += ball_velocity[0]
        ball[1] += ball_velocity[1]

        # Ball collision with top and bottom walls
        if ball[1] <= 0 or ball[1] >= HEIGHT - BALL_SIZE:
            ball_velocity[1] = -ball_velocity[1]

        # Ball collision with paddles
        if (ball[0] <= left_paddle[0] + PADDLE_WIDTH and left_paddle[1] < ball[1] < left_paddle[1] + PADDLE_HEIGHT) or \
           (ball[0] >= right_paddle[0] - BALL_SIZE and right_paddle[1] < ball[1] < right_paddle[1] + PADDLE_HEIGHT):
            ball_velocity[0] = -ball_velocity[0]

        # Scoring
        if ball[0] <= 0:
            right_score += 1
            ball = [WIDTH // 2 - BALL_SIZE // 2, HEIGHT // 2 - BALL_SIZE // 2]
            ball_velocity = [BALL_SPEED_X * random.choice((1, -1)), BALL_SPEED_Y * random.choice((1, -1))]
        elif ball[0] >= WIDTH:
            left_score += 1
            ball = [WIDTH // 2 - BALL_SIZE // 2, HEIGHT // 2 - BALL_SIZE // 2]
            ball_velocity = [BALL_SPEED_X * random.choice((1, -1)), BALL_SPEED_Y * random.choice((1, -1))]

        # Draw everything
        draw_window(left_paddle, right_paddle, ball, left_score, right_score)

    pygame.quit()

# Start the game
if __name__ == "__main__":
    main()
```

### 2D shooter game

```python
import pygame
import random

# Initialize pygame
pygame.init()

# Set up the game window
WIDTH, HEIGHT = 800, 600
window = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("2D Shooter")

# Set up colors
WHITE = (255, 255, 255)
BLACK = (0, 0, 0)
RED = (255, 0, 0)
GREEN = (0, 255, 0)

# Set up the game clock
clock = pygame.time.Clock()

# Player properties
PLAYER_WIDTH = 50
PLAYER_HEIGHT = 50
PLAYER_SPEED = 7
DASH_SPEED = 15
COOLDOWN = 1000

# Bullet properties
BULLET_WIDTH = 5
BULLET_HEIGHT = 10
BULLET_SPEED = 10

# Enemy properties
ENEMY_WIDTH = 50
ENEMY_HEIGHT = 50
ENEMY_SPEED = 5

# Font for score display
font = pygame.font.SysFont('arial', 30)

# Game variables
player_x = WIDTH // 2 - PLAYER_WIDTH // 2
player_y = HEIGHT - PLAYER_HEIGHT - 10
player_speed = 0
bullets = []
enemies = []
score = 0
last_dash_time = 0
is_dashing = False
dash_direction = 0  # 1 for right, -1 for left, 0 for no dash

# Function to draw the game window
def draw_window():
    window.fill(BLACK)
    
    # Draw the player
    pygame.draw.rect(window, GREEN, (player_x, player_y, PLAYER_WIDTH, PLAYER_HEIGHT))
    
    # Draw bullets
    for bullet in bullets:
        pygame.draw.rect(window, WHITE, pygame.Rect(bullet[0], bullet[1], BULLET_WIDTH, BULLET_HEIGHT))
    
    # Draw enemies
    for enemy in enemies:
        pygame.draw.rect(window, RED, pygame.Rect(enemy[0], enemy[1], ENEMY_WIDTH, ENEMY_HEIGHT))
    
    # Draw the score
    score_text = font.render(f"Score: {score}", True, WHITE)
    window.blit(score_text, (10, 10))
    
    pygame.display.update()

# Function to handle player movement
def handle_player_movement():
    # Get the current time in milliseconds
    current_time = pygame.time.get_ticks()
    
    global player_x
    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT] and player_x > 0:
        player_x -= PLAYER_SPEED
    if keys[pygame.K_RIGHT] and player_x < WIDTH - PLAYER_WIDTH:
        player_x += PLAYER_SPEED
        
# Dash cooldown function
def can_dash():
    global last_dash_time
    current_time = pygame.time.get_ticks()
    if current_time - last_dash_time >= DASH_COOLDOWN:
        last_dash_time = current_time
        return True
    return False

# Function to handle shooting
def shoot_bullet():
    bullet_x = player_x + PLAYER_WIDTH // 2 - BULLET_WIDTH // 2
    bullet_y = player_y
    bullets.append([bullet_x, bullet_y])

# Function to handle enemy movement
def move_enemies():
    global score
    for enemy in enemies:
        enemy[1] += ENEMY_SPEED
        if enemy[1] > HEIGHT:
            enemies.remove(enemy)
            score -= 1  # Penalize for enemies reaching the bottom
    
    # Add new enemies at random positions
    if random.random() < 0.02:
        enemy_x = random.randint(0, WIDTH - ENEMY_WIDTH)
        enemies.append([enemy_x, 0])

# Function to check for bullet-enemy collisions
def check_collisions():
    global score
    for bullet in bullets:
        for enemy in enemies:
            if pygame.Rect(bullet[0], bullet[1], BULLET_WIDTH, BULLET_HEIGHT).colliderect(pygame.Rect(enemy[0], enemy[1], ENEMY_WIDTH, ENEMY_HEIGHT)):
                bullets.remove(bullet)
                enemies.remove(enemy)
                score += 1  # Increase score for hitting an enemy
                break

# Main game loop
def main():
    global score
    global player_x
    global player_y
    global player_speed
    global bullets
    global enemies
    
    running = True
    while running:
        clock.tick(60)  # Set the frame rate to 60 FPS
        
        # Event handling
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False
            elif event.type == pygame.KEYDOWN:
                if event.key == pygame.K_SPACE:
                    shoot_bullet()
#             if event.key == pygame.K_a and can_dash():  # Dash left
#                 dash_direction = -1
#                 is_dashing = True
#             if event.key == pygame.K_d and can_dash():  # Dash right
#                 dash_direction = 1
#                 is_dashing = True

        # Handle dashing
        if is_dashing:
            player_x += dash_direction * DASH_SPEED
            is_dashing = False  # End the dash immediately
        
        # Handle player movement
        handle_player_movement()
        
        # Move the bullets
        for bullet in bullets[:]:
            bullet[1] -= BULLET_SPEED
            if bullet[1] < 0:
                bullets.remove(bullet)  # Remove bullets that go off-screen
        
        # Move the enemies
        move_enemies()
        
        # Check for collisions
        check_collisions()
        
        # Draw everything
        draw_window()
    
    pygame.quit()

# Start the game
if __name__ == "__main__":
    main()
```

```python
import pygame
import time

# Initialize pygame
pygame.init()

# Set up the screen
WIDTH, HEIGHT = 800, 600
window = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Shooter Game with Dash")

# Set up colors
WHITE = (255, 255, 255)
BLACK = (0, 0, 0)
GREEN = (0, 255, 0)

# Set up the clock
clock = pygame.time.Clock()

# Player properties
PLAYER_WIDTH = 50
PLAYER_HEIGHT = 50
PLAYER_SPEED = 5
DASH_SPEED = 15
DASH_COOLDOWN = 1000  # 1 second cooldown in milliseconds
last_dash_time = 0  # Last time the player dashed
player_x = WIDTH // 2 - PLAYER_WIDTH // 2
player_y = HEIGHT - PLAYER_HEIGHT - 20
is_dashing = False
dash_direction = 0  # 1 for right, -1 for left, 0 for no dash

# Bullet properties
BULLET_WIDTH = 5
BULLET_HEIGHT = 10
BULLET_SPEED = 10
bullets = []

# Font for score
font = pygame.font.SysFont('arial', 30)

# Game loop flag
running = True

# Dash cooldown function
def can_dash():
    global last_dash_time
    current_time = pygame.time.get_ticks()
    if current_time - last_dash_time >= DASH_COOLDOWN:
        last_dash_time = current_time
        return True
    return False

# Draw the player
def draw_player(x, y):
    pygame.draw.rect(window, GREEN, pygame.Rect(x, y, PLAYER_WIDTH, PLAYER_HEIGHT))

# Draw bullets
def draw_bullets():
    for bullet in bullets:
        pygame.draw.rect(window, WHITE, pygame.Rect(bullet[0], bullet[1], BULLET_WIDTH, BULLET_HEIGHT))

# Update bullets
def update_bullets():
    global bullets
    for bullet in bullets[:]:
        bullet[1] -= BULLET_SPEED
        if bullet[1] < 0:  # Remove bullet if it goes off-screen
            bullets.remove(bullet)

# Main game loop
while running:
    clock.tick(60)  # 60 frames per second

    # Event handling
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        if event.type == pygame.KEYDOWN:
            if event.key == pygame.K_LEFT and not is_dashing:
                player_x -= PLAYER_SPEED
            if event.key == pygame.K_RIGHT and not is_dashing:
                player_x += PLAYER_SPEED
            if event.key == pygame.K_SPACE:
                # Shoot a bullet
                bullet_x = player_x + PLAYER_WIDTH // 2 - BULLET_WIDTH // 2
                bullet_y = player_y
                bullets.append([bullet_x, bullet_y])
            if event.key == pygame.K_a and can_dash():  # Dash left
                dash_direction = -1
                is_dashing = True
            if event.key == pygame.K_d and can_dash():  # Dash right
                dash_direction = 1
                is_dashing = True

    # Handle dashing
    if is_dashing:
        player_x += dash_direction * DASH_SPEED
        is_dashing = False  # End the dash immediately

    # Prevent the player from going off the screen
    if player_x < 0:
        player_x = 0
    if player_x > WIDTH - PLAYER_WIDTH:
        player_x = WIDTH - PLAYER_WIDTH

    # Update bullets
    update_bullets()

    # Fill the screen with black
    window.fill(BLACK)

    # Draw the player
    draw_player(player_x, player_y)

    # Draw bullets
    draw_bullets()

    # Draw the score
    score_text = font.render("Press 'A' to Dash Left, 'D' to Dash Right", True, WHITE)
    window.blit(score_text, (10, 10))

    # Update the screen
    pygame.display.update()

# Quit Pygame
pygame.quit()
```
